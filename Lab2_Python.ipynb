```python
class Product:
    def __init__(self, calories_per_100g: float, weight_grams: float):
        self.calories_per_100g = calories_per_100g
        self.weight_grams = weight_grams
    def get_info(self) -> str:
        return (f"Калорийность 100 г: {self.calories_per_100g} ккал\n"
                f"Вес продукта: {self.weight_grams} г")
    def calc_total_calories(self) -> float:
        return self.calories_per_100g * self.weight_grams / 100.0
def read_float(prompt: str) -> float:
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Ошибка: значение не может быть отрицательным. Повторите ввод.")
                continue
            return value
        except ValueError:
            print("Ошибка: введите корректное число.")
def main():
    print("Расчёт калорийности продукта")
    results = []
    while True:
        raw = input("Калорийность 100 г продукта (ккал) [или 'q' для выхода]: ").strip()
        if raw.lower() == 'q':
            break
        try:
            calories = float(raw)
            if calories < 0:
                print("Ошибка: значение не может быть отрицательным.\n")
                continue
        except ValueError:
            print("Ошибка: введите корректное число.\n")
            continue
        weight = read_float("Вес продукта (г): ")
        product = Product(calories, weight)
        info = product.get_info()
        total = product.calc_total_calories()
        block = (
            info + "\n" +
            f"Общая калорийность: {total:.2f} ккал\n" +
            "-" * 40
        )
        results.append(block)
        print("\nРезультат")
        print(block)
        print()
    if results:
        print("\nВсе результаты")
        for block in results:
            print(block)
if __name__ == "__main__":
    main()
```

```python
class Time:
    def __init__(self, hours: int = 0, minutes: int = 0, seconds: int = 0):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds
        self._normalize()
    def _normalize(self):
        total = self.hours * 3600 + self.minutes * 60 + self.seconds
        total %= 24 * 3600
        self.hours = total // 3600
        self.minutes = (total % 3600) // 60
        self.seconds = total % 60
    def __del__(self):
        try:
            print(f"[Деструктор] Объект Time({self.hours}:{self.minutes}:{self.seconds}) уничтожен")
        except Exception:
            pass
    def total_seconds(self) -> int:
        return self.hours * 3600 + self.minutes * 60 + self.seconds
    def add_5_seconds(self):
        total = self.total_seconds() + 5
        total %= 24 * 3600
        self.hours = total // 3600
        self.minutes = (total % 3600) // 60
        self.seconds = total % 60
    def get_info(self) -> str:
        return (f"Время: {self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}\n"
                f"Всего секунд: {self.total_seconds()}")
def read_int(prompt: str, lo: int = 0, hi: int = None) -> int:
    while True:
        try:
            value = int(input(prompt))
            if value < lo or (hi is not None and value > hi):
                print(f"Ошибка: значение должно быть в диапазоне [{lo}..{hi}].")
                continue
            return value
        except ValueError:
            print("Ошибка: введите целое число.")
def main():
    print("Демонстрация работы класса Time\n")
    results = []
    print("1. Объект, созданный по умолчанию (0:0:0)")
    t_default = Time()
    results.append("Объект по умолчанию: \n" + t_default.get_info())
    print("\n2. Объект, созданный с параметрами (10, 25, 40)")
    t_const = Time(10, 25, 40)
    results.append("Объект с параметрами: \n" + t_const.get_info())
    print("\n3. Ввод значений полей с клавиатуры")
    hours = read_int("Введите часы (0..23): ", 0, 23)
    minutes = read_int("Введите минуты (0..59): ", 0, 59)
    seconds = read_int("Введите секунды (0..59): ", 0, 59)
    t_user = Time(hours, minutes, seconds)
    info_before = t_user.get_info()
    t_user.add_5_seconds()
    info_after = t_user.get_info()
    results.append(
        "Объект, введённый пользователем: \n"
        "Исходное состояние: \n" + info_before +
        "\nПосле увеличения на 5 секунд: \n" + info_after
    )
    print("\nРЕЗУЛЬТАТЫ (аналог Memo)")
    for block in results:
        print(block)
        print("-" * 45)
if __name__ == "__main__":
    main()
```

```python
# КЛАСС-РОДИТЕЛЬ
class Product:
    def __init__(self, calories_per_100g: float, weight_grams: float):
        self.calories_per_100g = calories_per_100g
        self.weight_grams = weight_grams
    def get_info(self) -> str:
        return (f"  Калорийность 100 г: {self.calories_per_100g} ккал\n"
                f"  Вес продукта: {self.weight_grams} г\n"
                f"  Общая калорийность: {self.calc_total_calories():.2f} ккал")
    def calc_total_calories(self) -> float:
        return self.calories_per_100g * self.weight_grams / 100.0
# КЛАСС-ПОТОМОК
class VitaminProduct(Product):
    def __init__(self, calories_per_100g: float,
                 weight_grams: float,
                 vitamin_c_per_gram: float):
        super().__init__(calories_per_100g, weight_grams)
        self.vitamin_c_per_gram = vitamin_c_per_gram
    def get_info(self) -> str:
        return (super().get_info() + "\n"
                f"  Витамин C в 1 г: {self.vitamin_c_per_gram} мг\n"
                f"  Всего витамина C: {self.calc_total_vitamin_c():.2f} мг")

    def calc_total_vitamin_c(self) -> float:
        return self.vitamin_c_per_gram * self.weight_grams
# Вспомогательные функции ввода
def read_float(prompt: str) -> float:
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Ошибка: значение не может быть отрицательным.")
                continue
            return value
        except ValueError:
            print("Ошибка: введите корректное число.")
# Демонстрация работы
def main():
    print("Демонстрация классов Product и VitaminProduct\n")
    results = []
    print("1. Класс-родитель (значения-константы)")
    parent_const = Product(250, 180)
    results.append("РОДИТЕЛЬ (константы):\n" + parent_const.get_info())
    print("2. Класс-потомок (значения-константы)")
    child_const = VitaminProduct(52, 200, 0.4)
    results.append("ПОТОМОК (константы):\n" + child_const.get_info())
    print("\n3. Ввод данных для класса-родителя")
    cal1 = read_float("  Калорийность 100 г (ккал): ")
    w1 = read_float("  Вес продукта (г): ")
    parent_user = Product(cal1, w1)
    results.append("РОДИТЕЛЬ (введён пользователем):\n" + parent_user.get_info())
    print("\n4. Ввод данных для класса-потомка")
    cal2 = read_float("  Калорийность 100 г (ккал): ")
    w2 = read_float("  Вес продукта (г): ")
    vit = read_float("  Витамин C в 1 г продукта (мг): ")
    child_user = VitaminProduct(cal2, w2, vit)
    results.append("ПОТОМОК (введён пользователем):\n" + child_user.get_info())
    print("\nРЕЗУЛЬТАТЫ")
    for block in results:
        print(block)
        print("-" * 45)
if __name__ == "__main__":
    main()
```

```python
# КЛАСС-РОДИТЕЛЬ: Время
class Time:
    def __init__(self, hours: int = 0, minutes: int = 0, seconds: int = 0):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds
        self._normalize()
    def _normalize(self):
        total = self.hours * 3600 + self.minutes * 60 + self.seconds
        total %= 24 * 3600
        self.hours = total // 3600
        self.minutes = (total % 3600) // 60
        self.seconds = total % 60
    def total_seconds(self) -> int:
        return self.hours * 3600 + self.minutes * 60 + self.seconds
    def get_info(self) -> str:
        return f"{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}"
# КЛАСС-ПОТОМОК: Расписание движения поездов
class TrainSchedule(Time):
    def __init__(self, train_number: str, direction: str,
                 hours: int = 0, minutes: int = 0, seconds: int = 0):
        super().__init__(hours, minutes, seconds)
        self.train_number = train_number
        self.direction = direction
    def minutes_until_departure(self, current: "Time") -> int:
        departure_sec = self.total_seconds()
        current_sec = current.total_seconds()
        diff = departure_sec - current_sec
        if diff < 0:
            diff += 24 * 3600
        return diff // 60
    def get_info(self) -> str:
        return (f"  Номер поезда: {self.train_number}\n"
                f"  Направление: {self.direction}\n"
                f"  Время отправления: {super().get_info()}")
def read_int(prompt: str, lo: int, hi: int) -> int:
    while True:
        try:
            value = int(input(prompt))
            if value < lo or value > hi:
                print(f"Ошибка: значение должно быть в диапазоне [{lo}..{hi}].")
                continue
            return value
        except ValueError:
            print("Ошибка: введите целое число.")
def read_str(prompt: str) -> str:
    while True:
        value = input(prompt).strip()
        if value:
            return value
        print("Ошибка: строка не может быть пустой.")
def main():
    print("Демонстрация классов Time и TrainSchedule\n")
    results = []
    parent_const = Time(10, 25, 40)
    results.append("РОДИТЕЛЬ (константы):\n"
                   f"  Время: {parent_const.get_info()}\n"
                   f"  Всего секунд: {parent_const.total_seconds()}")
    child_const = TrainSchedule("123А", "Москва — Санкт-Петербург", 18, 45, 0)
    current_const = Time(18, 30, 0)
    results.append("ПОТОМОК (константы):\n"
                   + child_const.get_info() + "\n"
                   f"  Текущее время: {current_const.get_info()}\n"
                   f"  До отправления: "
                   f"{child_const.minutes_until_departure(current_const)} мин")
    print("1. Ввод времени (класс Time)")
    h = read_int("  Часы (0..23): ", 0, 23)
    m = read_int("  Минуты (0..59): ", 0, 59)
    s = read_int("  Секунды (0..59): ", 0, 59)
    parent_user = Time(h, m, s)
    results.append("РОДИТЕЛЬ (введён пользователем):\n"
                   f"  Время: {parent_user.get_info()}\n"
                   f"  Всего секунд: {parent_user.total_seconds()}")
    print("\n2. Ввод данных о поезде (класс TrainSchedule)")
    num = read_str("  Номер поезда: ")
    direction = read_str("  Направление: ")
    dh = read_int("  Час отправления (0..23): ", 0, 23)
    dm = read_int("  Минуты отправления (0..59): ", 0, 59)
    child_user = TrainSchedule(num, direction, dh, dm, 0)
    print("\n3. Ввод текущего времени для расчёта")
    ch = read_int("  Текущий час (0..23): ", 0, 23)
    cm = read_int("  Текущие минуты (0..59): ", 0, 59)
    current_user = Time(ch, cm, 0)
    results.append("ПОТОМОК (введён пользователем):\n"
                   + child_user.get_info() + "\n"
                   f"  Текущее время: {current_user.get_info()}\n"
                   f"  До отправления: "
                   f"{child_user.minutes_until_departure(current_user)} мин")
    print("\nРЕЗУЛЬТАТЫ")
    for block in results:
        print(block)
        print("-" * 45)
if __name__ == "__main__":
    main()
```

```python
# КЛАСС 1-ГО УРОВНЯ: Кабель
class Cable:
    def __init__(self, cable_type: str, cores: int, diameter: float):
        self.cable_type = cable_type
        self.cores = cores
        self.diameter = diameter
    def quality(self) -> float:
        if self.cores == 0:
            return 0.0
        return self.diameter / self.cores
    def get_info(self) -> str:
        return (f"  Тип кабеля: {self.cable_type}\n"
                f"  Количество жил: {self.cores}\n"
                f"  Диаметр: {self.diameter} мм\n"
                f"  Q = диаметр / жилы = {self.quality():.4f}")
# КЛАСС 2-ГО УРОВНЯ: Кабель с оплёткой
class ShieldedCable(Cable):
    def __init__(self, cable_type: str, cores: int, diameter: float,
                 has_braid: bool):
        super().__init__(cable_type, cores, diameter)
        self.has_braid = has_braid
    def quality(self) -> float:
        q = super().quality()
        if self.has_braid:
            return 2.0 * q
        else:
            return 0.7 * q
    def get_info(self) -> str:
        braid = "есть" if self.has_braid else "нет"
        return (f"  Тип кабеля: {self.cable_type}\n"
                f"  Количество жил: {self.cores}\n"
                f"  Диаметр: {self.diameter} мм\n"
                f"  Наличие оплётки (P): {braid}\n"
                f"  Qp = {'2*Q' if self.has_braid else '0.7*Q'} = {self.quality():.4f}")
# Вспомогательные функции ввода
def read_int(prompt: str, lo: int = 1) -> int:
    while True:
        try:
            value = int(input(prompt))
            if value < lo:
                print(f"Ошибка: значение должно быть не меньше {lo}.")
                continue
            return value
        except ValueError:
            print("Ошибка: введите целое число.")
def read_float(prompt: str, lo: float = 0.0) -> float:
    while True:
        try:
            value = float(input(prompt))
            if value <= lo:
                print(f"Ошибка: значение должно быть больше {lo}.")
                continue
            return value
        except ValueError:
            print("Ошибка: введите число.")
def read_str(prompt: str) -> str:
    while True:
        value = input(prompt).strip()
        if value:
            return value
        print("Ошибка: строка не может быть пустой.")
def read_bool(prompt: str) -> bool:
    while True:
        raw = input(prompt).strip().lower()
        if raw in ("true", "да", "1", "yes", "y"):
            return True
        if raw in ("false", "нет", "0", "no", "n"):
            return False
        print("Ошибка: введите True/False (или да/нет, 1/0).")
def main():
    print("Демонстрация классов Cable и ShieldedCable\n")
    results = []
    c1 = Cable("ВВГ", 3, 6.0)
    results.append("КЛАСС 1-ГО УРОВНЯ (константы): \n" + c1.get_info())
    c2 = ShieldedCable("КВК", 4, 8.0, True)
    results.append("КЛАСС 2-ГО УРОВНЯ, P = True (константы): \n" + c2.get_info())
    c3 = ShieldedCable("КВК", 4, 8.0, False)
    results.append("КЛАСС 2-ГО УРОВНЯ, P = False (константы): \n" + c3.get_info())
    print("\n1. Ввод данных о кабеле (класс Cable)")
    t1 = read_str("  Тип кабеля: ")
    k1 = read_int("  Количество жил: ")
    d1 = read_float("  Диаметр (мм): ")
    user1 = Cable(t1, k1, d1)
    results.append("КЛАСС 1-ГО УРОВНЯ (введён пользователем): \n" + user1.get_info())
    print("\n2. Ввод данных о кабеле с оплёткой (класс ShieldedCable)")
    t2 = read_str("  Тип кабеля: ")
    k2 = read_int("  Количество жил: ")
    d2 = read_float("  Диаметр (мм): ")
    p = read_bool("  Наличие оплётки P (True/False): ")
    user2 = ShieldedCable(t2, k2, d2, p)
    results.append("КЛАСС 2-ГО УРОВНЯ (введён пользователем): \n" + user2.get_info())
    print("\nРЕЗУЛЬТАТЫ")
    for block in results:
        print(block)
        print("-" * 45)
if __name__ == "__main__":
    main()
```
